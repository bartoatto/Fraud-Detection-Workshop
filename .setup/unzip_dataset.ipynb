# Unzip Dataset Upload
One-time setup: extracts a zip you uploaded into the project's Domino Dataset, in place.

```python
import os
import zipfile

domino_username = os.environ.get("DOMINO_USER_NAME")
domino_project_name = os.environ.get("DOMINO_PROJECT_NAME", "my-local-project")
is_git_based_project = os.environ.get("DOMINO_IS_GIT_BASED")
domino_data_dir = os.environ.get("DOMINO_DATASETS_DIR", "/domino/datasets/")
if is_git_based_project == "true":
    domino_dataset_dir = f"{domino_data_dir}/{domino_project_name}/{domino_username}"
else:
    domino_dataset_dir = f"{domino_data_dir}/local/{domino_project_name}"

zip_filename = "fraud-workshop-data.zip"  # change if you named it differently
zip_path = f"{domino_dataset_dir}/{zip_filename}"
```

```python
if not os.path.exists(zip_path):
    raise FileNotFoundError(f"{zip_path} not found - check zip_filename above matches what you uploaded.")

with zipfile.ZipFile(zip_path, "r") as zf:
    zf.extractall(domino_dataset_dir)

print(f"Extracted {zip_filename} into {domino_dataset_dir}")
print(os.listdir(domino_dataset_dir))
```
